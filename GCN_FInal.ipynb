```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import f_oneway
from statsmodels.stats.multicomp import pairwise_tukeyhsd

# ============================
# 1. Load and prepare data
from google.colab import drive
drive.mount('/content/drive')

file_path = '/content/drive/MyDrive/Projects/Combined_PLV_Data.xlsx'

# ============================

# Change this to your actual file path
df = pd.read_excel(file_path) # or pd.read_excel('your_file.xlsx')

# Remove any completely empty columns
df = df.dropna(axis=1, how='all')

# List of brain region pairs (columns starting with 'R')
regions = [col for col in df.columns if col.startswith('R')]

# ============================
# 2. Compute variability across groups
# ============================

# Mean coherence per group
group_means = df.groupby('Class')[regions].mean()

# Standard deviation across groups (which regions differ the most in mean)
variability = group_means.std().sort_values(ascending=False)

# Top 10 most variable regions
most_variable_regions = variability.head(10).index.tolist()

print("Top 10 most variable regions:")
print(most_variable_regions)

# ============================
# 3. Perform ANOVA for each region
# ============================

significant_regions = []

for region in most_variable_regions:
    # Get coherence per class
    groups = [df.loc[df['Class'] == cls, region] for cls in df['Class'].unique()]
    f_stat, p_val = f_oneway(*groups)

    if p_val < 0.05:
        significant_regions.append((region, p_val))

# ============================
# 4. Report significant regions
# ============================

if not significant_regions:
    print("\nNo regions showed significant differences (p < 0.05).")
else:
    print("\nRegions with significant ANOVA differences (p < 0.05):")
    for region, p_val in sorted(significant_regions, key=lambda x: x[1]):
        print(f"{region}: p = {p_val:.4e}")

# ============================
# 5. Post-hoc Tukey tests and visualization
# ============================

# For the top 3 most significant regions
for region, p_val in sorted(significant_regions, key=lambda x: x[1])[:3]:
    print(f"\nPost-hoc Tukey test for {region}:")

    tukey = pairwise_tukeyhsd(
        endog=df[region],
        groups=df['Class'],
        alpha=0.05
    )
    print(tukey.summary())

    # Boxplot
    plt.figure(figsize=(8, 5))
    sns.boxplot(x='Class', y=region, data=df, palette='Set2')
    plt.title(f'{region} coherence across groups (ANOVA p = {p_val:.3e})')
    plt.ylabel('Coherence')
    plt.xlabel('Group')
    plt.tight_layout()
    plt.show()

    # Add pairwise significance markers
    # (Plotting significance bars manually)
    ax = sns.boxplot(x='Class', y=region, data=df, palette='Set2')
    y_max = df[region].max() * 1.1
    ax.set_ylim(top=y_max * 1.2)

    # Tukey results table for convenience
    results_table = tukey._results_table.data[1:]

    for i, result in enumerate(results_table):
        group1, group2 = result[0], result[1]
        p_adj = result[5]
        reject = result[6]
        if reject:
            x1 = list(df['Class'].unique()).index(group1)
            x2 = list(df['Class'].unique()).index(group2)
            # Draw bar
            plt.plot(
                [x1, x1, x2, x2],
                [y_max, y_max * 1.05, y_max * 1.05, y_max],
                lw=1.5, color='black'
            )
            plt.text(
                (x1 + x2) * 0.5,
                y_max * 1.08,
                '*',
                ha='center',
                va='bottom',
                color='black',
                fontsize=14
            )
    plt.title(f'Significant differences in {region}')
    plt.ylabel('Coherence')
    plt.xlabel('Group')
    plt.tight_layout()
    plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import sem  # standard error of the mean

# Define your group labels mapping
group_mapping = {
    'H': 'NC',
    'F': 'FD',
    'A': 'AZ'
}

# Colors for the groups (feel free to adjust)
group_colors = {
    'H': '#1f77b4',  # blue
    'F': '#ff7f0e',  # orange
    'A': '#ffcc00'   # yellow
}

# Sort significant regions by p-value
sorted_regions = sorted(significant_regions, key=lambda x: x[1])
region_names = [r for r, _ in sorted_regions]

# Collect data
means = {grp: [] for grp in group_mapping.keys()}
errors = {grp: [] for grp in group_mapping.keys()}
pvals = [p for _, p in sorted_regions]

# For each region, compute mean and SEM per group
for region in region_names:
    for grp in group_mapping.keys():
        data = df.loc[df['Class'] == grp, region]
        means[grp].append(data.mean())
        errors[grp].append(sem(data))

# Prepare bar positions
x = np.arange(len(region_names))
width = 0.25

fig, ax = plt.subplots(figsize=(16, 6))

# Plot bars for each group
bars = []
for i, grp in enumerate(group_mapping.keys()):
    bars.append(
        ax.bar(
            x + i * width,
            means[grp],
            width,
            yerr=errors[grp],
            capsize=5,
            label=group_mapping[grp],
            color=group_colors[grp],
            edgecolor='black'
        )
    )

# Add significance stars
for idx, p in enumerate(pvals):
    if p < 0.001:
        stars = '***'
    elif p < 0.01:
        stars = '**'
    elif p < 0.05:
        stars = '*'
    else:
        stars = ''
    if stars:
        ax.text(
            idx + width,
            max(means['H'][idx], means['F'][idx], means['A'][idx]) + 0.05,
            stars,
            ha='center',
            va='bottom',
            fontsize=14,
            color='black'
        )

# Style x-axis
ax.set_xticks(x + width)
ax.set_xticklabels(region_names, rotation=45, ha='right', fontsize=10)

ax.set_ylabel('Coherence')
ax.set_title('Candidate Biomarker Brain Connections Across Groups')

# Legend
ax.legend(title='Group')

# Grid and layout
ax.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import torch
from torch_geometric.data import Data, InMemoryDataset
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from google.colab import drive

drive.mount('/content/drive')

file_path = '/content/drive/MyDrive/Projects/Combined_PLV_Data.xlsx'

# Load data
df = pd.read_excel(file_path)

# List of top-10 significant regions
top_regions = [
    'R1195', 'R1207', 'R1004', 'R1215', 'R1186',
    'R1029', 'R1635', 'R1217', 'R991', 'R613'
]

# Convert your top regions to numeric, coercing errors
for col in top_regions:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Fill any NaNs with 0 (or you can choose mean or median imputation)
df[top_regions] = df[top_regions].fillna(0)

# Encode labels
label_encoder = LabelEncoder()
df['LabelEncoded'] = label_encoder.fit_transform(df['Class'])
print("Label mapping:", dict(zip(label_encoder.classes_, label_encoder.transform(label_encoder.classes_))))

# Define adjacency matrix: fully connected graph of 10 nodes
edge_index = torch.combinations(torch.arange(len(top_regions)), r=2).T

# Build PyTorch Geometric dataset
class EEGDataset(InMemoryDataset):
    def __init__(self, dataframe, transform=None):
        super().__init__('.', transform)
        self.data_list = []
        for i, row in dataframe.iterrows():
            x = torch.tensor(row[top_regions].values.astype(float), dtype=torch.float).unsqueeze(1)
            y = torch.tensor([row['LabelEncoded']], dtype=torch.long)
            data = Data(x=x, edge_index=edge_index, y=y)
            self.data_list.append(data)

    def len(self):
        return len(self.data_list)

    def get(self, idx):
        return self.data_list[idx]

# Set random seed for reproducibility
torch.manual_seed(42)
np.random.seed(42)

# First split: 70% training, 30% temp (will be split into val and test)
train_df, temp_df = train_test_split(df, test_size=0.3, random_state=42, stratify=df['LabelEncoded'])

# Second split: 50% of temp (15% of total) for validation, 50% (15% of total) for test
val_df, test_df = train_test_split(temp_df, test_size=0.5, random_state=42, stratify=temp_df['LabelEncoded'])

# Create datasets
train_dataset = EEGDataset(train_df)
val_dataset = EEGDataset(val_df)
test_dataset = EEGDataset(test_df)

print(f"Training samples: {len(train_dataset)}")
print(f"Validation samples: {len(val_dataset)}")
print(f"Test samples: {len(test_dataset)}")
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch_geometric.nn import GCNConv, global_mean_pool
from torch_geometric.data import DataLoader
from sklearn.metrics import confusion_matrix, classification_report
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

class ImprovedEEG_GCN(nn.Module):
    def __init__(self, num_features, num_classes):
        super().__init__()
        self.conv1 = GCNConv(num_features, 64)
        self.bn1 = nn.BatchNorm1d(64)
        self.conv2 = GCNConv(64, 128)
        self.bn2 = nn.BatchNorm1d(128)
        self.conv3 = GCNConv(128, 64)
        self.lin = nn.Linear(64, num_classes)
        self.dropout = nn.Dropout(0.3)

    def forward(self, data):
        x, edge_index, batch = data.x, data.edge_index, data.batch
        x = torch.relu(self.bn1(self.conv1(x, edge_index)))
        x = self.dropout(x)
        x = torch.relu(self.bn2(self.conv2(x, edge_index)))
        x = self.dropout(x)
        x = self.conv3(x, edge_index)
        x = global_mean_pool(x, batch)
        return self.lin(x)

def train_and_evaluate(model, train_loader, val_loader, test_loader, class_weights, epochs=50, patience=5):
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    model = model.to(device)

    # Enhanced optimizer with weight decay
    optimizer = optim.AdamW(model.parameters(), lr=0.001, weight_decay=1e-4)

    # Learning rate scheduler
    scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, 'min', patience=3, verbose=True)

    criterion = nn.CrossEntropyLoss(weight=class_weights.to(device))

    best_val_loss = float('inf')
    best_val_acc = 0
    counter = 0
    train_losses = []
    val_losses = []
    val_accuracies = []

    for epoch in range(epochs):
        # Training
        model.train()
        epoch_train_loss = 0
        for batch in train_loader:
            batch = batch.to(device)
            optimizer.zero_grad()
            out = model(batch)
            loss = criterion(out, batch.y.squeeze())
            loss.backward()

            # Gradient clipping
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)

            optimizer.step()
            epoch_train_loss += loss.item()

        # Validation
        model.eval()
        val_loss = 0
        correct = 0
        total = 0
        with torch.no_grad():
            for batch in val_loader:
                batch = batch.to(device)
                out = model(batch)
                val_loss += criterion(out, batch.y.squeeze()).item()
                pred = out.argmax(dim=1)
                correct += (pred == batch.y.squeeze()).sum().item()
                total += batch.y.size(0)

        # Calculate metrics
        epoch_train_loss /= len(train_loader)
        val_loss /= len(val_loader)
        val_acc = correct / total

        # Update scheduler
        scheduler.step(val_loss)

        # Store metrics for plotting
        train_losses.append(epoch_train_loss)
        val_losses.append(val_loss)
        val_accuracies.append(val_acc)

        # Early stopping and model checkpointing
        if val_loss < best_val_loss:
            best_val_loss = val_loss
            best_val_acc = val_acc
            counter = 0
            torch.save(model.state_dict(), 'best_model.pt')
        else:
            counter += 1
            if counter >= patience:
                print(f"Early stopping at epoch {epoch}")
                break

        print(f"Epoch {epoch+1}/{epochs} | "
              f"Train Loss: {epoch_train_loss:.4f} | "
              f"Val Loss: {val_loss:.4f} | "
              f"Val Acc: {val_acc:.4f} | "
              f"LR: {optimizer.param_groups[0]['lr']:.6f}")

    # Plot training curves
    plt.figure(figsize=(12, 4))
    plt.subplot(1, 2, 1)
    plt.plot(train_losses, label='Train Loss')
    plt.plot(val_losses, label='Val Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()

    plt.subplot(1, 2, 2)
    plt.plot(val_accuracies, label='Val Accuracy')
    plt.xlabel('Epoch')
    plt.ylabel('Accuracy')
    plt.legend()
    plt.tight_layout()
    plt.show()

    # Load best model and evaluate on test set
    model.load_state_dict(torch.load('best_model.pt'))
    model.eval()
    y_true, y_pred, y_probs = [], [], []
    with torch.no_grad():
        for batch in test_loader:
            batch = batch.to(device)
            out = model(batch)
            probs = torch.softmax(out, dim=1)
            pred = out.argmax(dim=1)
            y_true.extend(batch.y.squeeze().cpu().numpy())
            y_pred.extend(pred.cpu().numpy())
            y_probs.extend(probs.cpu().numpy())

    return y_true, y_pred, y_probs

def plot_confusion_matrix(y_true, y_pred, class_names):
    cm = confusion_matrix(y_true, y_pred)
    plt.figure(figsize=(8, 6))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=class_names, yticklabels=class_names)
    plt.title('Confusion Matrix')
    plt.ylabel('True Label')
    plt.xlabel('Predicted Label')
    plt.show()

def plot_classification_report(y_true, y_pred, class_names):
    report = classification_report(y_true, y_pred, target_names=class_names, output_dict=True)
    plt.figure(figsize=(8, 4))
    sns.heatmap(pd.DataFrame(report).iloc[:-1, :].T, annot=True, cmap='Blues')
    plt.title('Classification Report')
    plt.show()

if __name__ == "__main__":
    # Set random seeds for reproducibility
    torch.manual_seed(42)
    np.random.seed(42)

    # Get labels from the training dataset
    train_labels = [data.y.item() for data in train_dataset]

    # Add class weights to loss function
    class_counts = np.bincount(train_labels)
    class_weights = 1. / torch.tensor(class_counts, dtype=torch.float)

    # Handle potential infinities in class_weights
    class_weights[torch.isinf(class_weights)] = torch.median(class_weights[~torch.isinf(class_weights)])

    # Create data loaders
    train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)
    val_loader = DataLoader(val_dataset, batch_size=32)
    test_loader = DataLoader(test_dataset, batch_size=32)

    # Initialize and train model
    model = ImprovedEEG_GCN(
        num_features=1,
        num_classes=len(label_encoder.classes_)
    )

    y_true, y_pred, y_probs = train_and_evaluate(
        model, train_loader, val_loader, test_loader, class_weights
    )

    # Convert numeric labels back to original class names
    class_names = label_encoder.classes_
    y_true_names = label_encoder.inverse_transform(y_true)
    y_pred_names = label_encoder.inverse_transform(y_pred)

    # Plot confusion matrix
    plot_confusion_matrix(y_true, y_pred, class_names)

    # Plot enhanced classification report
    plot_classification_report(y_true_names, y_pred_names, class_names)

    # Print detailed classification report
    print("\nDetailed Classification Report:")
    print(classification_report(y_true_names, y_pred_names, target_names=class_names, digits=4))
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch_geometric.nn import GCNConv, global_mean_pool
from torch_geometric.data import DataLoader
from sklearn.metrics import confusion_matrix, classification_report
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from copy import deepcopy

class GaussianNoise(object):
    def __init__(self, mean=0., std=0.01):
        self.std = std
        self.mean = mean

    def __call__(self, data):
        # Create a deep copy to avoid modifying original data
        noisy_data = deepcopy(data)
        noise = torch.randn_like(noisy_data.x) * self.std + self.mean
        noisy_data.x = noisy_data.x + noise
        return noisy_data

class ImprovedEEG_GCN(nn.Module):
    def __init__(self, num_features, num_classes):
        super().__init__()
        self.conv1 = GCNConv(num_features, 64)
        self.bn1 = nn.BatchNorm1d(64)
        self.conv2 = GCNConv(64, 128)
        self.bn2 = nn.BatchNorm1d(128)
        self.conv3 = GCNConv(128, 64)
        self.lin = nn.Linear(64, num_classes)
        self.dropout = nn.Dropout(0.3)

    def forward(self, data):
        x, edge_index, batch = data.x, data.edge_index, data.batch
        x = torch.relu(self.bn1(self.conv1(x, edge_index)))
        x = self.dropout(x)
        x = torch.relu(self.bn2(self.conv2(x, edge_index)))
        x = self.dropout(x)
        x = self.conv3(x, edge_index)
        x = global_mean_pool(x, batch)
        return self.lin(x)

def train_and_evaluate(model, train_loader, val_loader, test_loader, class_weights, epochs=50, patience=5):
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    model = model.to(device)

    optimizer = optim.AdamW(model.parameters(), lr=0.001, weight_decay=1e-4)
    scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, 'min', patience=3, verbose=True)
    criterion = nn.CrossEntropyLoss(weight=class_weights.to(device))

    best_val_loss = float('inf')
    best_val_acc = 0
    counter = 0
    train_losses = []
    val_losses = []
    val_accuracies = []

    for epoch in range(epochs):
        # Training with Gaussian Noise augmentation
        model.train()
        epoch_train_loss = 0
        for batch in train_loader:
            batch = batch.to(device)

            # Apply Gaussian Noise to the batch
            noisy_batch = deepcopy(batch)
            noise = torch.randn_like(noisy_batch.x) * 0.02  # std=0.02
            noisy_batch.x = noisy_batch.x + noise

            optimizer.zero_grad()
            out = model(noisy_batch)
            loss = criterion(out, noisy_batch.y.squeeze())
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
            optimizer.step()
            epoch_train_loss += loss.item()

        # Validation (without noise)
        model.eval()
        val_loss = 0
        correct = 0
        total = 0
        with torch.no_grad():
            for batch in val_loader:
                batch = batch.to(device)
                out = model(batch)
                val_loss += criterion(out, batch.y.squeeze()).item()
                pred = out.argmax(dim=1)
                correct += (pred == batch.y.squeeze()).sum().item()
                total += batch.y.size(0)

        # Calculate metrics
        epoch_train_loss /= len(train_loader)
        val_loss /= len(val_loader)
        val_acc = correct / total

        scheduler.step(val_loss)
        train_losses.append(epoch_train_loss)
        val_losses.append(val_loss)
        val_accuracies.append(val_acc)

        if val_loss < best_val_loss:
            best_val_loss = val_loss
            best_val_acc = val_acc
            counter = 0
            torch.save(model.state_dict(), 'best_model.pt')
        else:
            counter += 1
            if counter >= patience:
                print(f"Early stopping at epoch {epoch}")
                break

        print(f"Epoch {epoch+1}/{epochs} | "
              f"Train Loss: {epoch_train_loss:.4f} | "
              f"Val Loss: {val_loss:.4f} | "
              f"Val Acc: {val_acc:.4f} | "
              f"LR: {optimizer.param_groups[0]['lr']:.6f}")

    # Plot training curves
    plt.figure(figsize=(12, 4))
    plt.subplot(1, 2, 1)
    plt.plot(train_losses, label='Train Loss')
    plt.plot(val_losses, label='Val Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()

    plt.subplot(1, 2, 2)
    plt.plot(val_accuracies, label='Val Accuracy')
    plt.xlabel('Epoch')
    plt.ylabel('Accuracy')
    plt.legend()
    plt.tight_layout()
    plt.show()

    # Evaluation
    model.load_state_dict(torch.load('best_model.pt'))
    model.eval()
    y_true, y_pred, y_probs = [], [], []
    with torch.no_grad():
        for batch in test_loader:
            batch = batch.to(device)
            out = model(batch)
            probs = torch.softmax(out, dim=1)
            pred = out.argmax(dim=1)
            y_true.extend(batch.y.squeeze().cpu().numpy())
            y_pred.extend(pred.cpu().numpy())
            y_probs.extend(probs.cpu().numpy())

    return y_true, y_pred, y_probs

def plot_confusion_matrix(y_true, y_pred, class_names):
    cm = confusion_matrix(y_true, y_pred)
    plt.figure(figsize=(8, 6))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=class_names, yticklabels=class_names)
    plt.title('Confusion Matrix with Gaussian Noise Augmentation')
    plt.ylabel('True Label')
    plt.xlabel('Predicted Label')
    plt.show()

if __name__ == "__main__":
    torch.manual_seed(42)
    np.random.seed(42)

    # Get class weights
    train_labels = [data.y.item() for data in train_dataset]
    class_counts = np.bincount(train_labels)
    class_weights = 1. / torch.tensor(class_counts, dtype=torch.float)
    class_weights[torch.isinf(class_weights)] = torch.median(class_weights[~torch.isinf(class_weights)])

    # Create data loaders
    train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)
    val_loader = DataLoader(val_dataset, batch_size=32)
    test_loader = DataLoader(test_dataset, batch_size=32)

    # Initialize model
    model = ImprovedEEG_GCN(
        num_features=1,
        num_classes=len(label_encoder.classes_)
    )

    # Train and evaluate
    y_true, y_pred, y_probs = train_and_evaluate(
        model, train_loader, val_loader, test_loader, class_weights
    )

    # Visualize results
    class_names = label_encoder.classes_
    y_true_names = label_encoder.inverse_transform(y_true)
    y_pred_names = label_encoder.inverse_transform(y_pred)

    plot_confusion_matrix(y_true, y_pred, class_names)

    print("\nClassification Report:")
    print(classification_report(y_true_names, y_pred_names, target_names=class_names, digits=4))
```
